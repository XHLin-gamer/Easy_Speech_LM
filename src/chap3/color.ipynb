```python
import matplotlib.pyplot as plt
import numpy as np

# Set up the figure and axis
fig, ax = plt.subplots(figsize=(10, 5), dpi=400)

# Define resolution width for the gradient bars
width = 1000
x = np.linspace(0, 1, width)

# 1. 1-bit Grayscale: 2^1 = 2 levels (Pure Black and White)
levels_1 = 2
band_1 = np.floor(x * levels_1) / (levels_1 - 1)

# 2. 2-bit Grayscale: 2^2 = 4 levels
levels_2 = 4
band_2 = np.floor(x * levels_2) / (levels_2 - 1)

# 3. 4-bit Grayscale: 2^4 = 16 levels
levels_4 = 16
band_4 = np.floor(x * levels_4) / (levels_4 - 1)

# 4. 8-bit Grayscale: 2^8 = 256 levels
levels_8 = 256
band_8 = np.floor(x * levels_8) / (levels_8 - 1)

# Group the bands and labels
bands = [band_1, band_2, band_4, band_8]
labels = [
    '1-bit\n(2 Levels)', 
    '2-bit\n(4 Levels)', 
    '4-bit\n(16 Levels)', 
    '8-bit\n(256 Levels)'
]

# Render each band as a horizontal grayscale image strip
for i, band in enumerate(bands):
    ax.imshow(band.reshape(1, width), cmap='gray', vmin=0, vmax=1, 
              extent=[0, 255, i - 0.35, i + 0.35], aspect='auto', origin='lower')

# Customize plot aesthetics
ax.set_yticks(range(4))
ax.set_yticklabels(labels, fontsize=11, fontweight='bold')
ax.set_xlim(0, 255)
ax.set_ylim(-0.8, 3.8)
ax.set_xlabel('Gray Value Intensity', fontsize=12, labelpad=10)
ax.set_title('BIT DEPTH QUANTIZATION: GRAYSCALE COMPARISON (1, 2, 4, 8 BIT)', fontsize=13, pad=15, fontweight='bold')

# Clean up axes spines
for spine in ['top', 'right', 'left']:
    ax.spines[spine].set_visible(False)
ax.tick_params(axis='y', length=0)　

plt.tight_layout()
plt.savefig("8bit.jpg",dpi=400)
```
