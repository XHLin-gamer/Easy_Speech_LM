```python
from pathlib import Path
import subprocess

def test_opus(input_wav: str, bitrates_kbps=[1, 1.1, 2, 4, 8, 16, 32]):
    src = Path(input_wav)
    out_dir = Path("opus_results")
    out_dir.mkdir(exist_ok=True)

    print(f"{'Bitrate':<10} | {'Compressed Size':<15} | Output WAV")
    print("-" * 50)

    for kbps in bitrates_kbps:
        opus_file = out_dir / f"{src.stem}_{kbps}k.opus"
        decoded_wav = out_dir / f"{src.stem}_{kbps}k_decoded.wav"

        # 1. Encode WAV -> Opus
        subprocess.run(
            [
                "ffmpeg",
                "-y",
                "-i",
                src,
                "-c:a",
                "libopus",
                "-b:a",
                f"{kbps}k",
                opus_file,
            ],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.DEVNULL,
        )

        # 2. Decode Opus -> WAV (for listening comparison)
        subprocess.run(
            ["ffmpeg", "-y", "-i", opus_file, decoded_wav],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.DEVNULL,
        )

        size_kb = opus_file.stat().st_size / 1024
        print(f"{kbps} kbps{'':<5} | {size_kb:6.1f} KB        | {decoded_wav.name}")
        
test_opus("./EMOTION100_001.wav")
```

```python
test_opus("./Morning.mp3")
```
